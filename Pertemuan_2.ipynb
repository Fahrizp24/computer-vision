2. Apakah perbedaan dan pengaruhnya pembuatan black image antara tipe data
int16 dan int32?

- int16 dan int32 sama-sama menampilkan gambar hitam kalau isinya nol.
Perbedaannya ada di rentang nilai piksel dan besar memori.
- int16 Nilai maksimum 32.767. lebih cepat kelihatan perubahannya (sedikit lebih terang).
- int32 Nilai maksimum jauh lebih besar (2 milyar). warnanya lebih gelap, butuh nilai jauh lebih besar untuk terlihat jelas.

3. Apakah kegunaan “google.colab.patches import cv2_imshow” pada potongan
kode berikut

- import cv2 as cv Digunakan untuk memanggil fungsi utama OpenCV (misalnya cv.imread, cv.cvtColor, cv.line, dll).
Ini adalah library inti OpenCV.

- sedangkan from google.colab.patches import cv2_imshow bukan bagian dari OpenCV, melainkan modul tambahan dari Google Colab. fungsinya untuk menampilkan hasilnya di Google Colab.

4. Apakah kegunaan “skimage import io” pada potongan kode soal nomor 3
- Potongan from skimage import io  merupakan modul io dari library scikit-image (skimage).
- Fungsinya: Untuk membaca (read) dan menulis (save) gambar dengan cara yang sederhana. Alternatif dari cv.imread() atau matplotlib.pyplot.imread().

```python
# 5. Buat 5 kotak berbagai ukuran dan warna random
import random

img_boxes = img.copy()
for i in range(5):
    x1, y1 = random.randint(0, img.shape[1]//2), random.randint(0, img.shape[0]//2)
    x2, y2 = x1 + random.randint(30,100), y1 + random.randint(30,100)
    color = (random.randint(0,255), random.randint(0,255), random.randint(0,255))
    cv.rectangle(img_boxes, (x1,y1), (x2,y2), color, 3)

plt.imshow(img_boxes)
plt.show()

```

```python
# 6. Tampilkan image dengan posisi terbalik
flip_img = cv.flip(img_boxes, -1)  # -1 = flip horizontal + vertical
plt.imshow(flip_img)
plt.show()
```
